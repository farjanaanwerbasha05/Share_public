#Lesson 4: Control flow (IF ELSE Statement)

```python
def age_calculation():
    age=int(input("Enter the age:"))
    if age < 18:
        print(f"Age :{age};You are not an adult")
    elif age >=18 and age <60:
        print(f"Age :{age};You are an adult ")
    else:
        print(f"Age :{age};You are senior citizen")
age_calculation()  # Calling the function
```

#Lesson 5: While and for loop

```python
def get_number():
        user_input=input("If you want to continue enter 'start', else 'stop'")
        numbers=[]
        if user_input.lower()== 'start':
            num_input=int(input("Howmany numbers do you want to enter?"))
            while len(numbers) < num_input:
                    user_entry = input(f"Enter a number {len(numbers) + 1}: ")
                    if user_entry.lower() == 'stop':
                         print("User chose to stop")  
                         break
                    try:
                        num=int(user_entry)
                        numbers.append(num)
                    except ValueError:
                        print(f"Entered Invalid value '{user_entry}' in {len(numbers)+1} position,please reenter a valid integer for {len(numbers)+1} position")
            print(f'*******Valid numbers*****\nEntered numbers:{sorted(numbers)}')
            print(f'Sum of all entered numbers:{sum(numbers)}')
            print(f'Average of entered numbers:{int(sum(numbers)/len(numbers))}')
        else:
            print(f"you entered {user_input}")

get_number()
                   
```
